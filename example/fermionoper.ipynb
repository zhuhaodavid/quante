## JW transformation

# 

```python
# JW transformation  spin -> fermion
import quante as qt

op = qt.generate.operas
builder = op.SpinBuilder()
L = 10
J, γ = 1, 0.0
for i in range(L-1):
    builder += "+-", [i+1, i], (J+γ)/2
    builder += "+-", [i, i+1], (J-γ)/2
ham = builder.build()

ham.jw_transfer()
```

```python
# JW transformation  fermion -> spin

import quante as qt

op = qt.generate.operas
builder = op.FermionBuilder()
L = 10
J, γ = 1, 0.0
for i in range(L-1):
    builder += "+-", [i+1, i], (J+γ)/2
    builder += "+-", [i, i+1], (J-γ)/2
ham = builder.build()

ham.jw_transfer()
```

```python
# 验证正确性：
import quante as qt
import numpy as np

op = qt.generate.operas

L = 5
basis = qt.generate.basis.spin_basis(L=L)
ham = qt.generate.operas.heisenberg_operator(L=L).expandxy(pauli=True)
mat1 = ham.to_matrix(basis)
ham = ham.jw_transfer()  # spin -> fermion
ham = ham.jw_transfer()  # fermion -> spin
mat2 = ham.to_matrix(basis)
print(np.linalg.norm(mat1 - mat2))
```

```python
# automata - fermion

# 验证正确性：
import quante as qt
import numpy as np

op = qt.generate.operas

L = 5
basis = qt.generate.basis.spin_basis(L=L)
ham = qt.generate.operas.heisenberg_operator(L=L).expandxy(pauli=True)
mat1 = ham.to_matrix(basis)
ham = ham.jw_transfer()  # spin -> fermion
ham = ham.jw_transfer()  # fermion -> spin
mpo = ham.to_mpo()
mat2 = mpo.to_matrix().numpy()
print(np.linalg.norm(mat1 - mat2))
print(mpo)
```

```python
import quante as qt
op = qt.generate.operas.fermion

L = 5
builder = op.SpinfulFermionBuilder()
for i in range(L-1):
    builder += "+-|", [i, i+1], -1.0
    builder += "-+|", [i, i+1], 1.0
    builder += "|+-", [i, i+1], -1.0
    builder += "|-+", [i, i+1], 1.0
for i in range(L):
    builder += "n|n", [i, i], 5.0
ham = builder.build()
ham
```

```python
import quante as qt
op = qt.generate.operas.fermion

L = 5
builder = op.SpinfulFermionBuilder()
for i in range(L-1):
    builder += "+-|", [i, i+1], -1.0
    builder += "-+|", [i, i+1], 1.0
    builder += "|+-", [i, i+1], -1.0
    builder += "|-+", [i, i+1], 1.0
for i in range(L):
    builder += "n|n", [i, i], 5.0
ham = builder.build()

basis = qt.generate.basis.quspin_spinful_fermion_basis(L=L)
mat = ham.to_matrix(basis, dtype=np.float64)
mat
```

```python
# 展成 spinless fermion
import quante as qt
op = qt.generate.operas.fermion

L = 5
builder = op.SpinfulFermionBuilder()
for i in range(L-1):
    builder += "+-|", [i, i+1], -1.0
    builder += "-+|", [i, i+1], 1.0
    builder += "|+-", [i, i+1], -1.0
    builder += "|-+", [i, i+1], 1.0
for i in range(L):
    builder += "n|n", [i, i], 5.0
ham = builder.build()
ham = ham.to_spinless(mode='extend')

basis = qt.generate.basis.quspin_spinless_fermion_basis(L=2*L)
mat = ham.to_matrix(basis, dtype=np.float64)
mat, np.linalg.eigvalsh(mat)[0]
```
