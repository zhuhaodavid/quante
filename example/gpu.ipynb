```python
# 多次 svd （张量网络中的主要情形）
import torch as tc
import time

mat = tc.randn(250, 250 ,dtype=tc.float64, device='cuda')
t = time.time()
for i in range(100):
    u, s, v = tc.linalg.svd(mat, full_matrices=False)
print("gpu time:", time.time() - t)


mat = mat.to('cpu')
t = time.time()
for i in range(100):
    u, s, v = tc.linalg.svd(mat, full_matrices=False)
print("cpu time:", time.time() - t)
```

```python
# 单次 svd
import torch as tc
import time

mat = tc.randn(2000, 2000 ,dtype=tc.float64, device='cuda')
t = time.time()
u, s, v = tc.linalg.svd(mat, full_matrices=False)
print("gpu time:", time.time() - t)


mat = mat.to('cpu')
t = time.time()
u, s, v = tc.linalg.svd(mat, full_matrices=False)
print("cpu time:", time.time() - t)
```

```python
# eigh
import torch as tc
import time

mat = tc.randn(4000, 4000 ,dtype=tc.float64, device='cuda')
mat = mat + mat.H
t = time.time()
v, u = tc.linalg.eigh(mat)
print("gpu time:", time.time() - t)


mat = mat.to('cpu')
t = time.time()
v, u = tc.linalg.eigh(mat)
print("cpu time:", time.time() - t)
```

```python
# 矩阵乘法
import torch as tc
import time

mat0 = tc.randn(10000, 10000, dtype=tc.float64, device='cuda')
mat = mat0.clone()
t = time.time()
mat2 = tc.matmul(mat, mat)
tc.cuda.synchronize()  # 等待GPU计算完成，否则时间会不准确
print("gpu time:", time.time() - t)


mat = mat0.clone().to('cpu')
t = time.time()
mat2 = tc.matmul(mat, mat)
print("cpu time:", time.time() - t)
```

```python
# 32 位矩阵乘法
import torch as tc
import time

mat0 = tc.randn(10000, 10000, dtype=tc.float32, device='cuda')
mat = mat0.clone()
t = time.time()
mat2 = tc.matmul(mat, mat)
tc.cuda.synchronize()  # 等待GPU计算完成，否则时间会不准确
print("gpu time:", time.time() - t)


mat = mat0.clone().to('cpu')
t = time.time()
mat2 = tc.matmul(mat, mat)
print("cpu time:", time.time() - t)
```
