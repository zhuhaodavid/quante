## GPU 效率

```python
# 多次 svd （张量网络中的主要情形）
import torch as tc
import time

mat = tc.randn(250, 250 ,dtype=tc.float64, device='cuda')
t = time.time()
for i in range(100):
    u, s, v = tc.linalg.svd(mat, full_matrices=False)
print("gpu time:", time.time() - t)


mat = mat.to('cpu')
t = time.time()
for i in range(100):
    u, s, v = tc.linalg.svd(mat, full_matrices=False)
print("cpu time:", time.time() - t)

# 北航工位电脑结果(py311)：
# gpu time: 12.930907011032104
# cpu time: 0.6241991519927979
```

```python
# 单次 svd
import torch as tc
import time

mat = tc.randn(4000, 4000 ,dtype=tc.float64, device='cuda')
t = time.time()
u, s, v = tc.linalg.svd(mat, full_matrices=False)
print("gpu time:", time.time() - t)


mat = mat.to('cpu')
t = time.time()
u, s, v = tc.linalg.svd(mat, full_matrices=False)
print("cpu time:", time.time() - t)

# 北航工位电脑结果(py311)：
# gpu time: 34.58361744880676
# cpu time: 17.55446147918701
```

```python
# eigh
import torch as tc
import time

mat = tc.randn(4000, 4000 ,dtype=tc.float64, device='cuda')
mat = mat + mat.H
t = time.time()
v, u = tc.linalg.eigh(mat)
tc.cuda.synchronize()  # 等待GPU计算完成，否则时间会不准确
print("gpu time:", time.time() - t)


mat = mat.to('cpu')
t = time.time()
v, u = tc.linalg.eigh(mat)
tc.cuda.synchronize()  # 等待GPU计算完成，否则时间会不准确
print("cpu time:", time.time() - t)

# 北航工位电脑结果(py311)：
# gpu time: 1.1364645957946777
# cpu time: 4.606389284133911
```

```python
# 矩阵乘法
import torch as tc
import time

mat0 = tc.randn(20000, 20000, dtype=tc.float64, device='cuda')
mat = mat0.clone()
t = time.time()
mat2 = tc.matmul(mat, mat)
tc.cuda.synchronize()  # 等待GPU计算完成，否则时间会不准确
print("双精度浮点 gpu:", time.time() - t)

mat = mat0.clone().to('cpu')
t = time.time()
mat2 = tc.matmul(mat, mat)
print("双精度浮点 cpu:", time.time() - t)

# 北航工位电脑结果(py311)：
# 双精度浮点 gpu: 32.479645013809204
# 双精度浮点 cpu: 42.61429738998413
```
