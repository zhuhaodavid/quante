```python
import time
import torch as tc

iternum = 10000
v = tc.rand(iternum, 3, dtype=tc.float64)

t = time.time()
sum1 = v[0,:].reshape(-1,1) @ v[0,:].reshape(1,-1)
for i in range(1,iternum):
    sum1 += v[i,:].reshape(-1,1) @ v[i,:].reshape(1,-1)
sum1 /= iternum
print(time.time() - t)

t = time.time()
sum2 = (v[:, :, None] @ v[:, None, :]).mean(0)
print(time.time() - t)

print(tc.allclose(sum1, sum2))
```
