```python
import quante as qt
import numpy as np

op = qt.generate.operas.spin

L = 14

builder = op.builder()
for i in range(L-1):
    builder += 'xx', [i, i+1], 1.
    builder += 'yy', [i, i+1], 1.
    builder += 'zz', [i, i+1], 1.
    builder += 'z', [i,], 10 * np.random.rand()
ham = builder.build()

basis = qt.generate.basis.spin_basis(L=L, Nup=L//2)
mat = ham.to_matrix(basis=basis)
eng = np.linalg.eigvalsh(mat)
qt.quantity.plot_level_spacing_distribution(
    eng, st_dists=['poisson', 'wigner-dyson-1']
)

import matplotlib.pyplot as plt
qt.basicfun.plt_style_use('science')
plt.legend()
plt.xlim(0,4)
```
